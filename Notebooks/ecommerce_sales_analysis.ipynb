### 5. Feature Engineering

```python
# high value orders 
df["High Value Order"] = np.where(df["Sales"] > 500, "Yes", "No")
```

```python
#profit margin
df["Profit Margin (%)"] = np.where(
    df["Sales"] != 0,
    (df["Profit"]/df["Sales"])*100,
    0
)
```

### 6. Business KPI

##### This section provides an overview of the company's overall business performance using key performance indicators (KPIs).

```python
plt.style.use("ggplot")
sns.set_theme()

df = pd.read_csv("../data/cleaned_sales.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
#Calculate business KPIs

total_revenue = df["Sales"].sum()

total_profit = df["Profit"].sum()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

average_order_value = total_revenue / total_orders

average_profit_margin = (total_profit / total_revenue) * 100
```

```python
print("-"*40)
print("E-COMMERCE SALES KPIs")
print("-"*40)

print(f"Total Revenue        : ${total_revenue:,.2f}")
print(f"Total Profit         : ${total_profit:,.2f}")
print(f"Total Orders         : {total_orders}")
print(f"Unique Customers     : {total_customers}")
print(f"Average Order Value  : ${average_order_value:,.2f}")
print(f"Profit Margin        : {average_profit_margin:.2f}%")
```

### 7. EDA and Visuaization

```python
#prepare monthly sales data
monthly_sales = (
    df.groupby(["Year", "Month Number", "Month"])["Sales"]
      .sum()
      .reset_index()
      .sort_values(["Year", "Month Number"])
)
```

```python
monthly_sales.head()
```

```python
#Plotting monthly revenue trend
plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=3
)

plt.title("Monthly Revenue Trend", fontsize=16, weight="bold")
plt.xlabel("Month")
plt.ylabel("Sales Revenue")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/monthly_sales_trend.png", dpi=300)

plt.show()
```

```python
#Finding best and  worst month
highest_month = monthly_sales.loc[
    monthly_sales["Sales"].idxmax()
]

lowest_month = monthly_sales.loc[
    monthly_sales["Sales"].idxmin()
]

print("Highest Sales Month")
print(highest_month)

print()

print("Lowest Sales Month")
print(lowest_month)
```

## Shipping Performance

##### Calculate Shipping days

```python
#convert shipping days
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
#create features
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
#check values
df["Shipping Days"].describe()
```

##### Shipping days distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x="Shipping Days",
    bins=15,
    kde=True
)

plt.title("Distribution of Shipping Days",
          fontsize=15,
          weight="bold")

plt.tight_layout()

plt.savefig("../images/shipping_days_distribution.png", dpi=300)

plt.show()
```
