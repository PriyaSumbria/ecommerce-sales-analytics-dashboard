```python
#extracting useful information
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month_name()
df["Month Number"] = df["Order Date"].dt.month
df["Day"] = df["Order Date"].dt.day_name()
```

### 5. Feature Engineering

```python
# high value orders 
df["High Value Order"] = np.where(df["Sales"] > 500, "Yes", "No")
```

```python
#profit margin
df["Profit Margin (%)"] = np.where(
    df["Sales"] != 0,
    (df["Profit"]/df["Sales"])*100,
    0
)
```

```python
#Saving the Clean Data
df.to_csv("../data/cleaned_sales.csv", index=False)
```

### 6. Business KPI

##### This section provides an overview of the company's overall business performance using key performance indicators (KPIs).

```python
plt.style.use("ggplot")
sns.set_theme()

df = pd.read_csv("../data/cleaned_sales.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
#Calculate business KPIs

total_revenue = df["Sales"].sum()

total_profit = df["Profit"].sum()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

average_order_value = total_revenue / total_orders

average_profit_margin = (total_profit / total_revenue) * 100
```

```python
print("-"*40)
print("E-COMMERCE SALES KPIs")
print("-"*40)

print(f"Total Revenue        : ${total_revenue:,.2f}")
print(f"Total Profit         : ${total_profit:,.2f}")
print(f"Total Orders         : {total_orders}")
print(f"Unique Customers     : {total_customers}")
print(f"Average Order Value  : ${average_order_value:,.2f}")
print(f"Profit Margin        : {average_profit_margin:.2f}%")
```

### 7. EDA and Visuaization

```python
#prepare monthly sales data
monthly_sales = (
    df.groupby(["Year", "Month Number", "Month"])["Sales"]
      .sum()
      .reset_index()
      .sort_values(["Year", "Month Number"])
)
```

```python
monthly_sales.head()
```

```python
#Plotting monthly revenue trend
plt.figure(figsize=(12,6))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=3
)

plt.title("Monthly Revenue Trend", fontsize=16, weight="bold")
plt.xlabel("Month")
plt.ylabel("Sales Revenue")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/monthly_sales_trend.png", dpi=300)

plt.show()
```

```python
#Finding best and  worst month
highest_month = monthly_sales.loc[
    monthly_sales["Sales"].idxmax()
]

lowest_month = monthly_sales.loc[
    monthly_sales["Sales"].idxmin()
]

print("Highest Sales Month")
print(highest_month)

print()

print("Lowest Sales Month")
print(lowest_month)
```

### Business Insights

```python
#preparing monthly profit data
monthly_profit = (
    df.groupby(["Year", "Month Number", "Month"])["Profit"]
      .sum()
      .reset_index()
      .sort_values(["Year", "Month Number"])
)
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=monthly_profit,
    x="Month",
    y="Profit",
    marker="o",
    linewidth=3
)

plt.title("Monthly Profit Trend", fontsize=16, weight="bold")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/monthly_profit_trend.png", dpi=300)

plt.show()
```

##### Sales By Category

```python
# Aggregate Category Sales
category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# creating bar chart
plt.figure(figsize=(10,6))

sns.barplot(
    data=category_sales,
    x="Category",
    y="Sales"
)

plt.title("Sales by Product Category", fontsize=16, weight="bold")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.tight_layout()

plt.savefig("../images/category_sales.png", dpi=300)

plt.show()
```

##### Profit By Category

```python
#aggregate profit
category_profit = (
    df.groupby("Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
#profit chart
plt.figure(figsize=(10,6))

sns.barplot(
    data=category_profit,
    x="Category",
    y="Profit"
)

plt.title("Profit by Product Category", fontsize=16, weight="bold")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.tight_layout()

plt.savefig("../images/category_profit.png", dpi=300)

plt.show()
```

```python
#Top 10 products by sales
top_products = (
    df.groupby("Sub-Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)
```

```python
#using horizontal bar chart
plt.figure(figsize=(11,6))

sns.barplot(
    data=top_products,
    y="Sub-Category",
    x="Sales"
)

plt.title("Top 10 Product Sub-Categories by Sales",
          fontsize=16,
          weight="bold")

plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.tight_layout()

plt.savefig("../images/top_products_sales.png", dpi=300)

plt.show()
```
